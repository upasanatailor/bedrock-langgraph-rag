```python
import os
import shutil
from langchain_aws import BedrockEmbeddings , ChatBedrock
from langchain_chroma import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Setup path
persist_directory = os.path.abspath("./chroma_db")

# 2. Cleanup (Optional: ONLY do this if you want to reset your DB)
# if os.path.exists(persist_directory):
#    shutil.rmtree(persist_directory)

# 3. Initialize Bedrock

import boto3
session = boto3.Session(profile_name="bedrock")
client = session.client("bedrock-runtime", region_name="eu-central-1")

embeddings = BedrockEmbeddings(
    model_id="amazon.titan-embed-text-v2:0",
    region_name="eu-central-1",
    client=client

)

loader = PyPDFLoader("./upasana_cv.pdf")
pdf_pages = loader.load()
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=100
)
pdf_chunks = text_splitter.split_documents(pdf_pages)

# 4. Initialize Chroma
try:
    vector_store = Chroma(
        collection_name="bedrock-collection",
        embedding_function=embeddings, 
        persist_directory=persist_directory
    )

    if pdf_chunks:
    # This adds the PDF segments and their metadata to the collection
        vector_store.add_documents(documents=pdf_chunks)
        print(f"✅ Successfully added {len(pdf_chunks)} PDF chunks to the vector store.")
    else:
         print("⚠️ No chunks found to add.")
    
except Exception as e:
    print(f"❌ Error still occurring: {e}")


# 2. Initialize the CHAT Model (The "Expert")
llm = ChatBedrock(
model_id="eu.amazon.nova-lite-v1:0",    client=client,
    model_kwargs={"temperature": 0.1}
)

```

```python
data = vector_store.get()

from langchain.chains import RetrievalQA

# 1. Turn your vector_store into a 'retriever'
# 'k=3' means it will find the 3 most relevant parts of your CV
retriever = vector_store.as_retriever(search_kwargs={"k": 3})

# 2. Create the QA Chain
qa_chain = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff", # 'Stuffs' all found context into one prompt
    retriever=retriever,
    return_source_documents=True # Optional: see which parts were used
)

# ✅ Remove the comma!
query = "What are the key skills mentioned in Upasana's CV?"

response = qa_chain.invoke({"query": query})
print(f"🤖 AI Response: {response['result']}")

print(f"🤖 AI Response: {response['result']}")
```
